## Download US ETF

### download_usetf list from etfdb

```python
key_symbols = f'us/etfs/symbols/etfdb'
```

```python
# HDF5 파일에서 데이터 읽기 및 병합
keys = [f'us/etfs/prices/yfinance', f'/us/snp500/prices/yfinance']
dfs = []

for key in keys:
    with pd.HDFStore(DATA_STORE, mode='r') as store:
        df = store[key]
        dfs.append(df.reset_index())

# 데이터프레임 병합
merged_df = pd.concat(dfs, ignore_index=True)
merged_df = merged_df.set_index(['ticker', 'date'])
# 병합된 데이터프레임 확인
print(merged_df.index.duplicated().mean())
print(merged_df.info())
```
